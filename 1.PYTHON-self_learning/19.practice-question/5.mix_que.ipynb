Do lists banao: Dono lists ke common elements find karo.

```python
list1 = [10, 20, 30, 40, 50]
list2 = [30, 40, 50, 60, 70]
list3 = [50, 60, 70, 80, 90]

common = []

for i in list1:
    if i in list2 and i in list3:
        common.append(i)

print(common)
```

Ek number input lo aur uske digits ka sum calculate karo.

```python
num = 4562
total = 0

while num > 0:
    digit = num % 10
    total = total + digit
    num = num // 10

print("sum of total digit :", total) 
```

Ek number input lo aur uska reverse print karo.

```python
num = 7895
reverse = 0

while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num = num // 10

print(reverse)
```

even_odd() function banao jo check kare ki number even hai ya odd.

```python
def even_odd(num):
    if num % 2 == 0:
        print("Even :", num)
    
    else:
        print("Odd :", num)

even_odd(10)
even_odd(5)
even_odd(9)
even_odd(70)
```

factorial() function banao jo kisi number ka factorial calculate kare.

```python
def factorial(num):
    fact = 1

    for i in range(1, num +1):
        fact = fact * i

    return fact

print(factorial(5))
```
